```python
# linear regression

# 1 input library
import numpy as np

import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression

# 2 input data
x = np.array([1,2,3,4,5]).reshape(-1,1)

# output data(scores)
y = np.array([50,55,65,70,80])

# 3 creating model {linear regression}

model=LinearRegression()

# 4 training the model

model.fit(x,y)

# 5 predict value

predicted=model.predict(x)

# 6 predict new data

new=model.predict([[6]])

# 7 visualization
plt.scatter(x,y)
plt.plot(x,predicted)
plt.show()
```

```python
#predict whether the student gonna fail or not

#LogisticRegression

# 1 import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# 2 input data
x = np.array([1,2,3,4,5]).reshape(-1,1)

# output data(scores)
y = np.array([0,0,0,1,1])

# 3 creating model {logistic regression}

model=LogisticRegression()

# 4 training the model

model.fit(x,y)

# 5 predict value

predicted=model.predict(x)

# 6 predict new data

new=model.predict([[2]])

# 7 visualization
plt.scatter(x,y)
plt.plot(x,predicted)
plt.show()
```
