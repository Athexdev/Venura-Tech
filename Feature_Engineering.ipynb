```python
#Feature engineering --->>its a very imp step in ml becacuse the quality of features that gonna directly affect to the model performance

#step 1 : handling missing data

import pandas as pd

data ={
    "Age" : [25, None, 30, 35]
}
df= pd.DataFrame(data)

df["Age"].fillna(df["Age"].mean(),inplace=True)
print(df)
```

```python
#Encoding categorical data
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
gender = ["male","female","male","female","male","male","female"]
encoded = le.fit_transform(gender)
print(encoded)
```

```python
#one hot encoding

import pandas as pd
df= pd.DataFrame({
    "color" : ["grey","red","blue","green","red","blue","orange","green","blue","red","orange","brown"]
})
encoded = pd.get_dummies(df)
print(encoded)
```

```python
#feature scaling


from sklearn.preprocessing import StandardScaler

import pandas as pd

df= pd.DataFrame({
    "Age": [20, 25,30,35,46,52],
    "Salary": [50000, 60000, 70000, 80000,90000,150000]
})

scaler = StandardScaler()
scaler = scaler.fit_transform(df)
print(scaler)
```
