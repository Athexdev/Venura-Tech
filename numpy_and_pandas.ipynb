```python
# matplotlib (used for data visualization)
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


x=np.linspace(0,10,100)
y=np.sin(x)

plt.plot(x, y, label='sin(x)' , color='red', linewidth=2)
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.legend()
plt.title('Sin Graph')
plt.grid(True)
plt.show()


```
