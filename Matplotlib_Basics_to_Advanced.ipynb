```python
# 2. BAR GRAPH
import matplotlib.pyplot as plt
students= ["A","B","C","D"]
marks = [80, 75, 90, 60]

plt.subplot(2, 3, 2)
plt.bar(students, marks)
plt.title('Student Marks')
plt.xlabel('Student')
plt.ylabel('Marks')

```

```python
# 3. HORIZONTAL BAR GRAPH
students = ['A', 'B', 'C', 'D']
marks = [80, 75, 90, 60]

plt.subplot(2, 3, 3)
plt.barh(students, marks)
plt.title('Student Marks (Horizontal)')
plt.xlabel('Marks')
plt.ylabel('Student')

```

```python
# 5. PIE CHART

marks=[40,30,20,10]
subjects =["Maths", "Science", "English", "History"]

plt.pie(marks, labels=subjects, autopct='%1.1f%%')
plt.title("Marks distribution")
plt.show()
```

```python
# 4. SCATTER PLOT

import matplotlib.pyplot as plt
a=[1,2,3,4]
b=[8,9,6,7]
plt.scatter(a,b)
plt.show()
```

```python
#histogram

import matplotlib.pyplot as plt

data =[10,20,20,30,40, 40, 40, 50]

plt.hist(data)

plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
#Area Plot(Fill Between)

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.figure(figsize=(8, 6))
plt.fill_between(x, y, alpha=0.5, color='blue')
plt.plot(x, y, color='darkblue', linewidth=2)

plt.title('Area Plot Example')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.grid(True)
plt.show()
```

```python
#Heatmap plot

import matplotlib.pyplot as plt
import numpy as np

data = np.random.rand(5, 5)

plt.imshow(data)
plt.colorbar()
plt.show()

```

```python
#3D plot Visualization

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')

x=[1,2,3]
y=[2,4,7]
z=[3,5,8]

ax.plot(x,y,z)
plt.show()
```
