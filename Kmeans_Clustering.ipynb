```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt


#sample data(x, y coordinates)

data = [
    [1,2],
    [1.5, 1.8],
    [5,8],
    [8, 8],
    [1, 0.6],
    [9, 11],
    [8,2],
    [10,2],
    [9,3]

]

#Create the k-mean Model

Kmeans = KMeans(n_clusters=3,
random_state=0)

#Train the model
Kmeans.fit(data)

#get cluster labels
labels= Kmeans.labels_

#get cluster centers
centers = Kmeans.cluster_centers_
print("cluster labels:")
print(labels)

print("cluster centers:")
print(centers)

#plot data points
for i in range(len(data)):
  plt.scatter(data[i][0], data[i][1])

  #plot the centroids
  plt.scatter(
      centers[:,0],
      centers[:,1],
      c='red',
      marker='x',
      s=100
  )

#plot the data points
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Create sample dataset
X, y = make_blobs(
    n_samples=150,
    centers=3,
    cluster_std=0.4,
    random_state=42
)

# Apply K-Means
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

# Get centroids
centroids = kmeans.cluster_centers_

# Plot Cluster 1
plt.scatter(
    X[labels == 0, 0],
    X[labels == 0, 1],
    s=100,
    c='red',
    label='Cluster 1'
)

# Plot Cluster 2
plt.scatter(
    X[labels == 1, 0],
    X[labels == 1, 1],
    s=100,
    c='blue',
    label='Cluster 2'
)

# Plot Cluster 3
plt.scatter(
    X[labels == 2, 0],
    X[labels == 2, 1],
    s=100,
    c='green',
    label='Cluster 3'
)

# Plot Centroids
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    s=400,
    c='yellow',
    label='Centroids'
)

plt.title("K-Means Clustering")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```
